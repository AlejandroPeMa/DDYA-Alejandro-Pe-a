## 7) Análisis de complejidad **[15 pts]**

### Justificación

El programa utiliza dos algoritmos principales. Primero, **Merge Sort** divide la lista en dos partes de manera recursiva y después mezcla las partes ordenadas. En cada nivel de la recursión se procesan aproximadamente `n` elementos y existen `log2(n)` niveles, por lo que su complejidad es `O(n log n)` tanto en el mejor como en el peor caso.

Después de ordenar, se utiliza **Búsqueda Binaria**. En cada iteración se descarta aproximadamente la mitad de los elementos posibles. Por esto, en el peor caso se necesitan `log2(n)` comparaciones y su complejidad es `O(log n)`. En el mejor caso, el elemento buscado está justo en la posición central y solamente se necesita una comparación, por lo que es `O(1)`.

### Función T[n]

Para Merge Sort, la recurrencia es:

`T(n) = 2T(n/2) + O(n)`

Al resolverla se obtiene:

`T(n) = O(n log n)`

Para Búsqueda Binaria:

`T(n) = T(n/2) + O(1)`

Por tanto:

`T(n) = O(log n)`

Si se considera el programa completo, primero se debe ordenar y después realizar la búsqueda. Entonces:

`T(n) = O(n log n) + O(log n)`

Como `n log n` crece más rápido que `log n`, la complejidad dominante del programa completo es:

**T(n) = O(n log n)**

### Mejor y peor caso

| Parte del programa | Mejor caso | Peor caso |
|---|---|---|
| Merge Sort | O(n log n) | O(n log n) |
| Búsqueda Binaria | O(1) | O(log n) |
| Programa completo | O(n log n) | O(n log n) |

La diferencia entre el mejor y el peor caso del programa completo no cambia el orden de crecimiento porque Merge Sort domina el costo total. La Búsqueda Binaria sí cambia entre `O(1)` y `O(log n)`, pero ambas son menores que `O(n log n)`.

```python
# Ejercicio: Organización y búsqueda de productos en una tienda


# Función para ordenar con Merge Sort
def merge_sort(lista):
    # Caso base: si la lista tiene 0 o 1 elemento, ya está "ordenada"
    if len(lista) <= 1:
        return lista

    # Dividimos la lista en dos mitades
    mitad = len(lista) // 2
    izquierda = lista[:mitad]
    derecha = lista[mitad:]

    # Ordenamos cada mitad por separado
    izquierda = merge_sort(izquierda)
    derecha = merge_sort(derecha)

    # Unimos las dos mitades ya ordenadas
    return mezclar(izquierda, derecha)


def mezclar(izq, der):
    resultado = []
    i = 0
    j = 0

    # Comparamos los elementos de ambas listas
    while i < len(izq) and j < len(der):
        if izq[i] < der[j]:
            resultado.append(izq[i])
            i = i + 1
        else:
            resultado.append(der[j])
            j = j + 1

    # Agregamos los elementos restantes de la izquierda
    while i < len(izq):
        resultado.append(izq[i])
        i = i + 1

    # Agregamos los elementos restantes de la derecha
    while j < len(der):
        resultado.append(der[j])
        j = j + 1

    return resultado


# Función para buscar con Búsqueda Binaria
def busqueda_binaria(lista, buscado):
    inicio = 0
    fin = len(lista) - 1

    while inicio <= fin:
        medio = (inicio + fin) // 2

        if lista[medio] == buscado:
            return True

        elif lista[medio] < buscado:
            inicio = medio + 1

        else:
            fin = medio - 1

    return False


# Programa principal

n = int(input("Ingrese la cantidad de productos: "))

codigos = []

for i in range(n):
    codigo = int(input(f"Ingrese el código del producto {i + 1}: "))
    codigos.append(codigo)

print("Lista original:", codigos)

codigos_ordenados = merge_sort(codigos)

print("Lista ordenada:", codigos_ordenados)

buscado = int(input("Ingrese el código que desea buscar: "))

if busqueda_binaria(codigos_ordenados, buscado):
    print(f"El producto con código {buscado} fue encontrado.")
else:
    print(f"El producto con código {buscado} no se encuentra registrado.")
```
