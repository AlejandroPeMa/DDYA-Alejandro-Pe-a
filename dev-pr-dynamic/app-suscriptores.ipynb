## 1) Descripción del problema **[1 pts]**                                                                              
Una plataforma de streaming está analizando el crecimiento de nuevos suscriptores.
Se ha observado que el número de nuevos suscriptores en el mes n sigue el siguiente
comportamiento:
• En el mes 0 se registraron 0 nuevos usuarios.
• En el mes 1 se registró 1 nuevo usuario.
• A partir del mes 2, el número de nuevos usuarios es igual a la suma de los nuevos usuarios de
los dos meses anteriores, debido al efecto combinado de referidos del mes anterior y
recomendaciones acumuladas del mes previo.
Relación de recurrencia:

T(n) = 0 si n = 0

T(n) = 1 si n = 1

T(n) = T(n-1) + T(n-2) si n ≥ 2

Determinar con memoización cuántos nuevos suscriptores habrá en el mes n.

## Justificación:
La version recursiva sin memoizacion de T(n)=T(n-1) + T(n-2) recalcula los mismos subproblemas una y otra vez. Lo que produce una complejidad temporal O(2^n) ese seria nuestro peor caso.

Con memoizacion cada valor de n entre 0 y N se calcula una unica vez y asi no tener que volver hacerlo, si el usuario vuelve a ingresar el mismo n que ingreso antes se resuelve ingresando a memo. Como hay N+1 subproblemas distintos y cada uno se calcula en tiempo constante es decir por medio de una suma, la complejidad es 0(n). la complejidad espacial tambien es O(n) tanto por la tabla de memoizacion y tambien por la recursión.

## Funcion T(n):

T(n)={ 0 si n=0, 1 si n=1, T(n-1)+T(n-2) si n>=2}

## Mejor y peor caso:


*   Mejor caso: n=0 o n=1 el algoritmo retorna inmediatamente y tambien si n ha sido calculado antes, con complejidad O(1).

*   Peor Caso: n grande y sin ningun dato en memo. el algoritmo debe resolver n+1 subproblemas antes de llegar al resultado final con complejidad O(n) gracias a la memoización.



## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def suscriptores(n, memo=None):
   if not isinstance(n, int) or n < 0:
       raise ValueError("n debe ser un entero no negativo")
   if memo is None:
       memo = {}
   if n == 0:
       return 0
   if n == 1:
       return 1
   if n in memo:
       return memo[n]
   memo[n] = suscriptores(n - 1, memo) + suscriptores(n - 2, memo)
   return memo[n]

n = int(input("Ingrese el mes n: "))
print(f"Nuevos suscriptores en el mes {n}: {suscriptores(n)}")
```
