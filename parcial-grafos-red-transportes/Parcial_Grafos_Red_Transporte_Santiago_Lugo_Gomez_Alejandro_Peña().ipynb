## 1) Descripción del problema **[1 pts]**

Una empresa de transportes urbano quiere encontrar la forma de mostrar las conexiones entre las estaciones de sus red. Cada estacion es un vetrtice y cada conexion directa entre dos estaciones es una arista no dirigida, porque si en dado caso que viajaramos de Central a Norte podemos ir de Norte a Central.

El sistema debe mostrar las conexiones de cada estacion, tambien recorrer la red con BFS y DFS desde cualquier estacion, encontrar la ruta con menos conexiones entre dos estaciones y regitsrar una nueva estacion.

##Análisis de complejidad
Para saber qué tan rápido funciona el programa tenemos en cuenta dos cosas:

V: Es la cantidad de estaciones que tenemos.

E: Es la cantidad de conexiones entre las estaciones.

Como las conexiones funcionan en ambos sentidos, cada una se guarda dos veces.

##Complejidad de las funciones
Agregar conexión:

Revisa que las estaciones no estén conectadas antes de agregar una nueva conexión. El tiempo depende de cuántas conexiones tenga cada estación.

**Mostrar red:** Pasa por todas las estaciones y muestra con cuáles están conectadas. Su complejidad es O(V + E).

**BFS:** Va pasando por las estaciones usando una cola. Cada estación se visita una sola vez y se revisan sus conexiones. Su complejidad es O(V + E).

**DFS:** Hace algo parecido a BFS pero sigue un camino hasta donde pueda antes de regresar. También visita cada estación una sola vez. Su complejidad es O(V + E).

**Buscar ruta:** Utiliza BFS para encontrar el camino con menos conexiones entre dos estaciones. Su complejidad es O(V + E).

**Agregar estación:** Permite crear una estación y conectarla con otras que ya existen. El tiempo depende de cuántas conexiones se quieran agregar.

**Función T(n) de BFS**

Para calcular el tiempo que tarda BFS tenemos en cuenta que primero se prepara la cola y se guardan las estaciones visitadas.

Después el programa va pasando por cada estación y revisa sus conexiones.
La fórmula queda así:

T(V, E) = c₁ + c₂·V + 2c₃·E

Esto quiere decir que el tiempo depende de la cantidad de estaciones y conexiones que tenga la red.

Por eso su complejidad es O(V + E).

En DFS pasa algo parecido porque también se revisan las estaciones y sus conexiones sin repetir las visitas.

**Mejor caso**

El mejor caso se da cuando la red tiene pocas conexiones, por ejemplo cuando las estaciones están conectadas una detrás de otra.
En este caso el programa tiene menos conexiones que revisar y su complejidad es O(n).

Para buscar una ruta el caso más rápido sería que la estación de inicio y la de destino fueran la misma.

**Peor caso**

El peor caso sería cuando todas las estaciones están conectadas entre sí.
Esto hace que el programa tenga muchas más conexiones que revisar y su complejidad puede llegar a ser O(n²).

**Conclusión**

En nuestro parcial tenemos 7 estaciones y 7 conexiones. Como no todas las estaciones están conectadas entre sí, la red es bastante sencilla.

Por eso BFS y DFS pueden recorrerla sin tener que revisar demasiadas conexiones.

En la gráfica podemos comparar cómo cambian las operaciones y el tiempo cuando la red tiene pocas o muchas conexiones.

```python
import time
import matplotlib.pyplot as plt
from collections import deque


def bfs_contando(grafo, inicio):
    operaciones = 0
    visitados = {inicio}
    cola = deque([inicio])
    while cola:
        actual = cola.popleft()
        operaciones += 1
        for vecino in grafo[actual]:
            operaciones += 1
            if vecino not in visitados:
                visitados.add(vecino)
                cola.append(vecino)
    return operaciones


def red_linea(n):
    grafo = {i: [] for i in range(n)}
    for i in range(n - 1):
        grafo[i].append(i + 1)
        grafo[i + 1].append(i)
    return grafo


def red_completa(n):
    return {i: [j for j in range(n) if j != i] for i in range(n)}


tamanos = list(range(10, 401, 30))
ops_mejor, ops_peor, t_mejor, t_peor = [], [], [], []
for n in tamanos:
    linea = red_linea(n)
    completa = red_completa(n)
    ops_mejor.append(bfs_contando(linea, 0))
    ops_peor.append(bfs_contando(completa, 0))
    inicio = time.perf_counter()
    for _ in range(20):
        bfs_contando(linea, 0)
    t_mejor.append((time.perf_counter() - inicio) / 20 * 1000)
    inicio = time.perf_counter()
    for _ in range(20):
        bfs_contando(completa, 0)
    t_peor.append((time.perf_counter() - inicio) / 20 * 1000)

fig, ejes = plt.subplots(1, 2, figsize=(13, 4.8))
ejes[0].plot(tamanos, ops_mejor, "o-", color="#2e7d32", label="Mejor caso: línea, E = n - 1, O(n)")
ejes[0].plot(tamanos, ops_peor, "s-", color="#c62828", label="Peor caso: completa, E = n(n-1)/2, O(n²)")
ejes[0].set_title("Operaciones de BFS según el número de estaciones", color="black")
ejes[0].set_xlabel("n (vértices)")
ejes[0].set_ylabel("Operaciones contadas")
ejes[0].grid(alpha=0.3)
ejes[1].plot(tamanos, t_mejor, "o-", color="#2e7d32", label="Mejor caso: línea")
ejes[1].plot(tamanos, t_peor, "s-", color="#c62828", label="Peor caso: completa")
ejes[1].set_title("Tiempo medido de BFS", color="black")
ejes[1].set_xlabel("n (vértices)")
ejes[1].set_ylabel("Tiempo (ms)")
ejes[1].grid(alpha=0.3)
ejes[0].legend(loc="upper center", bbox_to_anchor=(0.5, -0.16), frameon=False)
ejes[1].legend(loc="upper center", bbox_to_anchor=(0.5, -0.16), frameon=False)
plt.tight_layout()
plt.show()

for n, a, b in zip(tamanos[::4], ops_mejor[::4], ops_peor[::4]):
    print(f"n = {n:3d}   mejor caso = {a:6d} operaciones   peor caso = {b:7d} operaciones")
```

Con n = 400, el mejor caso hace unas 1.200 operaciones y el peor caso unas 160.000: la curva verde crece como una recta (O(n)) y la roja como una parábola (O(n²)). Esto confirma que el costo real de BFS depende de **V + E**: en una red completa E crece con n², y eso domina el tiempo.

## 6) Código funcional **[15 pts]**

```python
from collections import deque
import matplotlib.pyplot as plt

POSICIONES = {
    "Central": (0, 3),
    "Norte": (-1.2, 2),
    "Museo": (1.2, 2),
    "Parque": (-2.2, 1),
    "Universidad": (-0.2, 1),
    "Hospital": (2.0, 1),
    "Terminal": (0.9, 0)
}
DESFASE = (1.3, -0.9)


def crear_red_inicial():
    red = {}
    for estacion in ["Central", "Norte", "Museo", "Parque", "Universidad", "Hospital", "Terminal"]:
        agregar_estacion(red, estacion)
    for a, b in [
        ("Central", "Norte"),
        ("Central", "Museo"),
        ("Norte", "Parque"),
        ("Norte", "Universidad"),
        ("Museo", "Hospital"),
        ("Universidad", "Terminal"),
        ("Hospital", "Terminal")
    ]:
        agregar_conexion(red, a, b)
    return red


def agregar_estacion(red, nombre):
    if nombre not in red:
        red[nombre] = []


def agregar_conexion(red, a, b):
    if a == b or a not in red or b not in red:
        return False
    if b in red[a]:
        return False
    red[a].append(b)
    red[b].append(a)
    return True


def mostrar_red(red):
    for nombre in red:
        print(nombre + ":", ", ".join(red[nombre]))


def bfs(red, inicio):
    if inicio not in red:
        return None, None
    visitados = {inicio}
    cola = deque([inicio])
    orden = []
    nivel = {inicio: 0}
    while cola:
        actual = cola.popleft()
        orden.append(actual)
        for vecino in red[actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                nivel[vecino] = nivel[actual] + 1
                cola.append(vecino)
    return orden, nivel


def dfs(red, inicio):
    if inicio not in red:
        return None
    visitados = set()
    orden = []
    explorar(red, inicio, visitados, orden)
    return orden


def explorar(red, actual, visitados, orden):
    visitados.add(actual)
    orden.append(actual)
    for vecino in red[actual]:
        if vecino not in visitados:
            explorar(red, vecino, visitados, orden)


def buscar_ruta(red, origen, destino):
    if origen not in red or destino not in red:
        return None
    anterior = {origen: None}
    cola = deque([origen])
    while cola:
        actual = cola.popleft()
        if actual == destino:
            break
        for vecino in red[actual]:
            if vecino not in anterior:
                anterior[vecino] = actual
                cola.append(vecino)
    if destino not in anterior:
        return []
    ruta = []
    paso = destino
    while paso is not None:
        ruta.append(paso)
        paso = anterior[paso]
    ruta.reverse()
    return ruta


def registrar_estacion(red, nombre, vecinas):
    if nombre == "" or nombre in red:
        return "Error: la estación ya existe o el nombre está vacío."
    vecinas = list(dict.fromkeys(vecinas))
    for vecina in vecinas:
        if vecina not in red:
            return "Error: " + vecina + " no existe en la red."
    if len(vecinas) < 2:
        return "Error: debe conectarse con al menos dos estaciones existentes."
    agregar_estacion(red, nombre)
    for vecina in vecinas:
        agregar_conexion(red, nombre, vecina)
    return "Estación " + nombre + " registrada con " + str(len(vecinas)) + " conexiones."


def imprimir_bfs(red, inicio):
    orden, nivel = bfs(red, inicio)
    if orden is None:
        print("Error: la estación " + inicio + " no existe en la red.")
        return
    print("Recorrido BFS desde " + inicio + ": " + " -> ".join(orden))
    for n in range(max(nivel.values()) + 1):
        print("  Nivel " + str(n) + ": " + ", ".join(x for x in orden if nivel[x] == n))


def imprimir_dfs(red, inicio):
    orden = dfs(red, inicio)
    if orden is None:
        print("Error: la estación " + inicio + " no existe en la red.")
        return
    print("Recorrido DFS desde " + inicio + ": " + " -> ".join(orden))


def imprimir_ruta(red, origen, destino):
    ruta = buscar_ruta(red, origen, destino)
    if ruta is None:
        print("Error: la estación de origen o de destino no existe en la red.")
    elif ruta == []:
        print("No existe una ruta entre " + origen + " y " + destino + ".")
    else:
        print("Ruta encontrada: " + " -> ".join(ruta))
        print("Cantidad de conexiones: " + str(len(ruta) - 1))
    return ruta


def dibujar_red(red, ruta=None, titulo="Red"):
    pos = dict(POSICIONES)
    for nombre in red:
        if nombre not in pos:
            conocidos = [pos[x] for x in red[nombre] if x in pos]
            x = sum(p[0] for p in conocidos) / len(conocidos) + DESFASE[0]
            y = sum(p[1] for p in conocidos) / len(conocidos) + DESFASE[1]
            pos[nombre] = (x, y)
    ruta = ruta or []
    en_ruta = set(zip(ruta, ruta[1:])) | set(zip(ruta[1:], ruta))
    fig, ax = plt.subplots(figsize=(8, 5.5))
    for a in red:
        for b in red[a]:
            if a < b:
                color = "#e65100" if (a, b) in en_ruta else "#9e9e9e"
                grosor = 3.5 if (a, b) in en_ruta else 1.6
                ax.plot([pos[a][0], pos[b][0]], [pos[a][1], pos[b][1]], color=color, lw=grosor, zorder=1)
    for nombre in red:
        if ruta and nombre == ruta[0]:
            fondo, borde = "#c8e6c9", "#2e7d32"
        elif ruta and nombre == ruta[-1]:
            fondo, borde = "#ffcdd2", "#c62828"
        elif nombre in ruta:
            fondo, borde = "#ffe0b2", "#e65100"
        elif nombre not in POSICIONES:
            fondo, borde = "#e1bee7", "#7b1fa2"
        else:
            fondo, borde = "#dae8fc", "#1565c0"
        ax.text(pos[nombre][0], pos[nombre][1], nombre, ha="center", va="center", fontsize=11, zorder=2,
                bbox=dict(boxstyle="round,pad=0.45", fc=fondo, ec=borde, lw=1.8))
    ax.set_title(titulo, fontsize=13, color="black")
    ax.set_xlim(-3.0, 3.6)
    ax.set_ylim(-0.9, 3.5)
    ax.axis("off")
    plt.show()


red = crear_red_inicial()
print("Red cargada:", len(red), "estaciones")
```

## 7) Tests aplicados **[15 pts]**
Cada test arma una red nueva con `crear_red_inicial()`, para que un test no dependa de otro.

```python
resultados = []


def probar(nombre, condicion):
    resultados.append(condicion)
    print(("OK    " if condicion else "FALLA ") + nombre)


r = crear_red_inicial()
probar("La red inicial tiene 7 estaciones", len(r) == 7)
probar("La red tiene 7 conexiones (suma de listas = 2E)", sum(len(x) for x in r.values()) == 14)
probar("Grafo no dirigido: cada conexión está en ambos sentidos", all(a in r[b] for a in r for b in r[a]))
probar("No se agrega una conexión duplicada", agregar_conexion(r, "Central", "Norte") is False and r["Central"].count("Norte") == 1)
probar("No se agrega la conexión repetida al revés", agregar_conexion(r, "Norte", "Central") is False)
probar("No se permite un lazo (estación consigo misma)", agregar_conexion(r, "Central", "Central") is False)

orden, nivel = bfs(r, "Central")
probar("BFS desde Central da el orden esperado", orden == ["Central", "Norte", "Museo", "Parque", "Universidad", "Hospital", "Terminal"])
probar("BFS calcula bien el nivel de cada estación", nivel == {"Central": 0, "Norte": 1, "Museo": 1, "Parque": 2, "Universidad": 2, "Hospital": 2, "Terminal": 3})
probar("BFS visita cada estación una sola vez", len(orden) == len(set(orden)) == len(r))
probar("BFS con estación inexistente devuelve None", bfs(r, "Estadio") == (None, None))

orden = dfs(r, "Central")
probar("DFS desde Central da el orden esperado", orden == ["Central", "Norte", "Parque", "Universidad", "Terminal", "Hospital", "Museo"])
probar("DFS visita cada estación una sola vez", len(orden) == len(set(orden)) == len(r))
probar("DFS avanza en profundidad: el segundo visitado es vecino del primero", orden[1] in r[orden[0]])
probar("DFS con estación inexistente devuelve None", dfs(r, "Estadio") is None)
probar("BFS y DFS desde Universidad visitan todas las estaciones", set(bfs(r, "Universidad")[0]) == set(dfs(r, "Universidad")) == set(r))

probar("Ruta Central -> Terminal es la esperada", buscar_ruta(r, "Central", "Terminal") == ["Central", "Norte", "Universidad", "Terminal"])
probar("La ruta tiene 3 conexiones", len(buscar_ruta(r, "Central", "Terminal")) - 1 == 3)
probar("La ruta de BFS es la mínima (igual al nivel del destino)", len(buscar_ruta(r, "Central", "Terminal")) - 1 == bfs(r, "Central")[1]["Terminal"])
probar("Ruta de una estación a sí misma tiene 0 conexiones", buscar_ruta(r, "Central", "Central") == ["Central"])
probar("Cada paso de la ruta es una conexión real", all(b in r[a] for a, b in zip(buscar_ruta(r, "Parque", "Hospital"), buscar_ruta(r, "Parque", "Hospital")[1:])))
probar("Ruta con estación inexistente devuelve None", buscar_ruta(r, "Central", "Estadio") is None)
agregar_estacion(r, "Bodega")
probar("Ruta sin conexión devuelve lista vacía", buscar_ruta(r, "Central", "Bodega") == [])

r = crear_red_inicial()
probar("Rechaza registro con una sola conexión", registrar_estacion(r, "Aeropuerto", ["Terminal"]).startswith("Error") and "Aeropuerto" not in r)
probar("Rechaza registro con una estación inexistente", registrar_estacion(r, "Aeropuerto", ["Terminal", "Estadio"]).startswith("Error") and "Aeropuerto" not in r)
probar("Rechaza la misma estación repetida como dos conexiones", registrar_estacion(r, "Aeropuerto", ["Terminal", "Terminal"]).startswith("Error"))
probar("Registra Aeropuerto con Terminal y Hospital", not registrar_estacion(r, "Aeropuerto", ["Terminal", "Hospital"]).startswith("Error"))
probar("Las conexiones nuevas quedan en ambos sentidos", r["Aeropuerto"] == ["Terminal", "Hospital"] and "Aeropuerto" in r["Terminal"] and "Aeropuerto" in r["Hospital"])
probar("No permite registrar Aeropuerto dos veces", registrar_estacion(r, "Aeropuerto", ["Terminal", "Hospital"]).startswith("Error"))
probar("Después del registro BFS llega a Aeropuerto", "Aeropuerto" in bfs(r, "Central")[0])
probar("Después del registro DFS llega a Aeropuerto", "Aeropuerto" in dfs(r, "Central"))
probar("Después del registro hay ruta Central -> Aeropuerto", buscar_ruta(r, "Central", "Aeropuerto")[-1] == "Aeropuerto")

print()
print(str(sum(resultados)) + " de " + str(len(resultados)) + " tests pasaron")
assert all(resultados)
```
