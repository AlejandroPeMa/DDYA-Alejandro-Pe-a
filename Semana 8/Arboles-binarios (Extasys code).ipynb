### 5) Análisis de complejidad **[15 pts]**

### Justificación

En un BST, una inserción recorre desde la raíz hasta encontrar la posición correcta. Si el árbol está balanceado, una inserción cuesta aproximadamente O(log n). En el peor caso, el árbol puede quedar degenerado y una inserción cuesta O(n).

Por eso, al construir el BST:
- Mejor caso: T(n) = O(n log n)
- Peor caso: T(n) = O(n²)

En un AVL la altura se mantiene en O(log n), por lo que cada inserción cuesta O(log n), incluyendo el balanceo y las rotaciones. Construir el AVL cuesta O(n log n).

Los recorridos visitan cada nodo una vez:
- Preorden: T(n) = O(n)
- Inorden: T(n) = O(n)
- Postorden: T(n) = O(n)

### Función T[n]
- BST peor caso: T(n) = T(n-1) + c → O(n²)
- BST balanceado: T(n) ≈ T(n/2) + c → O(n log n)
- AVL: T(n) = O(log n) por inserción → O(n log n)
- Recorridos: T(n) = O(n)


```python
# Gráfica de complejidad
import numpy as np
import matplotlib.pyplot as plt

n_valores = np.arange(1, 21)

bst_mejor = n_valores * np.log2(n_valores + 1)
bst_peor = n_valores ** 2
avl = n_valores * np.log2(n_valores + 1)

plt.figure(figsize=(9, 5))
plt.plot(n_valores, bst_mejor, label="BST - mejor O(n log n)")
plt.plot(n_valores, bst_peor, label="BST - peor O(n²)")
plt.plot(n_valores, avl, label="AVL O(n log n)")
plt.xlabel("Cantidad de elementos n")
plt.ylabel("Operaciones aproximadas")
plt.title("Comparación de complejidad")
plt.legend()
plt.grid(True)
plt.show()
```

## 6) Código funcional **[15 pts]**

La siguiente implementación construye el BST y el AVL manualmente, realiza los tres recorridos de forma recursiva y genera las gráficas de los árboles.


```python
# ==========================================================
# CÓDIGO FUNCIONAL - BST Y AVL
# ==========================================================

import matplotlib.pyplot as plt


# -----------------------------
# Nodo
# -----------------------------
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izquierda = None
        self.derecha = None
        self.altura = 1


# ==========================================================
# BST
# ==========================================================

def insertar_bst(raiz, valor):
    if raiz is None:
        return Nodo(valor)

    if valor < raiz.valor:
        raiz.izquierda = insertar_bst(raiz.izquierda, valor)
    elif valor > raiz.valor:
        raiz.derecha = insertar_bst(raiz.derecha, valor)

    return raiz


# ==========================================================
# AVL
# ==========================================================

def altura(nodo):
    if nodo is None:
        return 0
    return nodo.altura


def actualizar_altura(nodo):
    nodo.altura = 1 + max(altura(nodo.izquierda),
                           altura(nodo.derecha))


def factor_balance(nodo):
    if nodo is None:
        return 0
    return altura(nodo.izquierda) - altura(nodo.derecha)


# Rotación simple a la derecha
def rotacion_derecha(y):
    x = y.izquierda
    temporal = x.derecha

    x.derecha = y
    y.izquierda = temporal

    actualizar_altura(y)
    actualizar_altura(x)

    return x


# Rotación simple a la izquierda
def rotacion_izquierda(x):
    y = x.derecha
    temporal = y.izquierda

    y.izquierda = x
    x.derecha = temporal

    actualizar_altura(x)
    actualizar_altura(y)

    return y


def insertar_avl(raiz, valor):
    if raiz is None:
        return Nodo(valor)

    if valor < raiz.valor:
        raiz.izquierda = insertar_avl(raiz.izquierda, valor)
    elif valor > raiz.valor:
        raiz.derecha = insertar_avl(raiz.derecha, valor)
    else:
        return raiz

    actualizar_altura(raiz)

    balance = factor_balance(raiz)

    # Caso izquierda-izquierda
    if balance > 1 and valor < raiz.izquierda.valor:
        return rotacion_derecha(raiz)

    # Caso derecha-derecha
    if balance < -1 and valor > raiz.derecha.valor:
        return rotacion_izquierda(raiz)

    # Caso izquierda-derecha
    if balance > 1 and valor > raiz.izquierda.valor:
        raiz.izquierda = rotacion_izquierda(raiz.izquierda)
        return rotacion_derecha(raiz)

    # Caso derecha-izquierda
    if balance < -1 and valor < raiz.derecha.valor:
        raiz.derecha = rotacion_derecha(raiz.derecha)
        return rotacion_izquierda(raiz)

    return raiz


# ==========================================================
# RECORRIDOS RECURSIVOS
# ==========================================================

def preorden(raiz, resultado=None):
    if resultado is None:
        resultado = []

    if raiz is not None:
        resultado.append(raiz.valor)
        preorden(raiz.izquierda, resultado)
        preorden(raiz.derecha, resultado)

    return resultado


def inorden(raiz, resultado=None):
    if resultado is None:
        resultado = []

    if raiz is not None:
        inorden(raiz.izquierda, resultado)
        resultado.append(raiz.valor)
        inorden(raiz.derecha, resultado)

    return resultado


def postorden(raiz, resultado=None):
    if resultado is None:
        resultado = []

    if raiz is not None:
        postorden(raiz.izquierda, resultado)
        postorden(raiz.derecha, resultado)
        resultado.append(raiz.valor)

    return resultado


# ==========================================================
# VISUALIZACIÓN
# ==========================================================

def posiciones_arbol(raiz):
    posiciones = {}
    contador = [0]

    def recorrer(nodo, nivel):
        if nodo is None:
            return

        recorrer(nodo.izquierda, nivel + 1)

        posiciones[nodo] = (contador[0], -nivel)
        contador[0] += 1

        recorrer(nodo.derecha, nivel + 1)

    recorrer(raiz, 0)

    return posiciones


def dibujar_arbol(raiz, titulo):
    if raiz is None:
        print("El árbol está vacío.")
        return

    posiciones = posiciones_arbol(raiz)

    plt.figure(figsize=(12, 7))

    def dibujar_conexiones(nodo):
        if nodo is None:
            return

        x, y = posiciones[nodo]

        if nodo.izquierda is not None:
            xi, yi = posiciones[nodo.izquierda]
            plt.plot([x, xi], [y, yi], linewidth=1.5)
            dibujar_conexiones(nodo.izquierda)

        if nodo.derecha is not None:
            xd, yd = posiciones[nodo.derecha]
            plt.plot([x, xd], [y, yd], linewidth=1.5)
            dibujar_conexiones(nodo.derecha)

    dibujar_conexiones(raiz)

    for nodo, (x, y) in posiciones.items():
        plt.scatter(x, y, s=900, zorder=3)
        plt.text(x, y, str(nodo.valor),
                 ha="center", va="center",
                 fontsize=12, color="white",
                 fontweight="bold")

    plt.title(titulo)
    plt.axis("off")
    plt.show()


# Mostrar factores de balance del AVL
def mostrar_balance(raiz):
    if raiz is None:
        return

    mostrar_balance(raiz.izquierda)

    print("Nodo:", raiz.valor,
          "| Altura:", altura(raiz),
          "| Factor de balance:", factor_balance(raiz))

    mostrar_balance(raiz.derecha)


# ==========================================================
# INGRESO DE DATOS
# ==========================================================

n = int(input("Ingrese la cantidad de productos: "))

codigos = []

for i in range(n):
    codigo = int(input(f"Ingrese el código {i + 1}: "))
    codigos.append(codigo)

print("\nCódigos ingresados:")
print(codigos)


# ==========================================================
# CONSTRUCCIÓN DEL BST Y AVL
# ==========================================================

raiz_bst = None
raiz_avl = None

for codigo in codigos:
    raiz_bst = insertar_bst(raiz_bst, codigo)
    raiz_avl = insertar_avl(raiz_avl, codigo)


# ==========================================================
# MOSTRAR ÁRBOLES
# ==========================================================

print("\nÁRBOL BINARIO DE BÚSQUEDA (BST)")
dibujar_arbol(raiz_bst, "Árbol Binario de Búsqueda (BST)")

print("\nÁRBOL AVL")
dibujar_arbol(raiz_avl, "Árbol AVL")


# ==========================================================
# FACTORES DE BALANCE
# ==========================================================

print("\nFACTORES DE BALANCE DEL AVL")
mostrar_balance(raiz_avl)


# ==========================================================
# RECORRIDOS
# ==========================================================

print("\nPREORDEN")
print("BST:", preorden(raiz_bst))
print("AVL:", preorden(raiz_avl))

print("\nINORDEN")
print("BST:", inorden(raiz_bst))
print("AVL:", inorden(raiz_avl))

print("\nPOSTORDEN")
print("BST:", postorden(raiz_bst))
print("AVL:", postorden(raiz_avl))


# ==========================================================
# COMPARACIÓN
# ==========================================================

print("\nCOMPARACIÓN")
print("Altura BST:", altura(raiz_bst))
print("Altura AVL:", altura(raiz_avl))

print("\nCaracterística especial del inorden:")
print("En un BST, el recorrido inorden muestra los valores ordenados de menor a mayor.")
```

## 7) Tests aplicados **[15 pts]**

Se utilizan los datos del ejemplo del enunciado para comprobar que los árboles y los recorridos funcionan correctamente.


```python
# ==========================================================
# TESTS
# ==========================================================

datos_test = [30, 20, 10, 25, 28, 27, 5, 50]

bst_test = None
avl_test = None

for codigo in datos_test:
    bst_test = insertar_bst(bst_test, codigo)
    avl_test = insertar_avl(avl_test, codigo)

esperado = [5, 10, 20, 25, 27, 28, 30, 50]

# El inorden debe quedar ordenado
assert inorden(bst_test) == esperado
assert inorden(avl_test) == esperado

# Ambos árboles tienen 8 nodos
assert len(preorden(bst_test)) == 8
assert len(preorden(avl_test)) == 8

# Todos los recorridos contienen los mismos valores
assert sorted(preorden(bst_test)) == esperado
assert sorted(postorden(bst_test)) == esperado
assert sorted(preorden(avl_test)) == esperado
assert sorted(postorden(avl_test)) == esperado


# Comprobar que el AVL está balanceado
def avl_balanceado(raiz):
    if raiz is None:
        return True

    balance = factor_balance(raiz)

    if balance < -1 or balance > 1:
        return False

    return avl_balanceado(raiz.izquierda) and avl_balanceado(raiz.derecha)


assert avl_balanceado(avl_test)

print("Todos los tests fueron aprobados correctamente.")
```

### Pruebas de rendimiento

Además de comprobar que los árboles funcionan, se mide cuánto tardan en construirse y en recorrerse a medida que crece la cantidad de códigos. Se usan dos tipos de datos: códigos en orden aleatorio y códigos que llegan ya ordenados, que es el peor caso del BST.

```python
# ==========================================================
# PRUEBAS DE RENDIMIENTO: CONSTRUCCIÓN
# ==========================================================

import gc
import math
import random
import sys
import time
import matplotlib.pyplot as plt

sys.setrecursionlimit(10000)
random.seed(7)


def altura_arbol(raiz):
    if raiz is None:
        return 0
    return 1 + max(altura_arbol(raiz.izquierda), altura_arbol(raiz.derecha))


def construir(datos, insertar):
    raiz = None
    for valor in datos:
        raiz = insertar(raiz, valor)
    return raiz


def medir(funcion, *args):
    mejor = float("inf")
    for _ in range(5):
        gc.disable()
        inicio = time.perf_counter()
        resultado = funcion(*args)
        fin = time.perf_counter()
        gc.enable()
        mejor = min(mejor, fin - inicio)
    return mejor * 1000, resultado


tamanos = list(range(100, 1501, 100))

t_bst_aleatorio, t_avl_aleatorio = [], []
t_bst_ordenado, t_avl_ordenado = [], []
h_bst_aleatorio, h_avl_aleatorio = [], []
h_bst_ordenado, h_avl_ordenado = [], []

for n in tamanos:
    aleatorios = random.sample(range(1, n * 10), n)
    ordenados = sorted(aleatorios)

    t, arbol = medir(construir, aleatorios, insertar_bst)
    t_bst_aleatorio.append(t)
    h_bst_aleatorio.append(altura_arbol(arbol))

    t, arbol = medir(construir, aleatorios, insertar_avl)
    t_avl_aleatorio.append(t)
    h_avl_aleatorio.append(altura_arbol(arbol))

    t, arbol = medir(construir, ordenados, insertar_bst)
    t_bst_ordenado.append(t)
    h_bst_ordenado.append(altura_arbol(arbol))

    t, arbol = medir(construir, ordenados, insertar_avl)
    t_avl_ordenado.append(t)
    h_avl_ordenado.append(altura_arbol(arbol))

fig, ejes = plt.subplots(1, 2, figsize=(13, 5))

ejes[0].plot(tamanos, t_bst_aleatorio, marker="o", label="BST")
ejes[0].plot(tamanos, t_avl_aleatorio, marker="o", label="AVL")
ejes[0].set_title("Construcción con códigos aleatorios")

ejes[1].plot(tamanos, t_bst_ordenado, marker="o", label="BST")
ejes[1].plot(tamanos, t_avl_ordenado, marker="o", label="AVL")
ejes[1].set_title("Construcción con códigos ordenados (peor caso del BST)")

for eje in ejes:
    eje.set_xlabel("Cantidad de códigos n")
    eje.set_ylabel("Tiempo (ms)")
    eje.grid(True)
    eje.legend()

plt.tight_layout()
plt.show()

print(f"Con n = {tamanos[-1]}:")
print(f"BST aleatorio: {t_bst_aleatorio[-1]:.2f} ms | AVL aleatorio: {t_avl_aleatorio[-1]:.2f} ms")
print(f"BST ordenado: {t_bst_ordenado[-1]:.2f} ms | AVL ordenado: {t_avl_ordenado[-1]:.2f} ms")
```

```python
# ==========================================================
# PRUEBAS DE RENDIMIENTO: ALTURA DE LOS ÁRBOLES
# ==========================================================

referencia = [math.log2(n) for n in tamanos]

fig, ejes = plt.subplots(1, 2, figsize=(13, 5))

ejes[0].plot(tamanos, h_bst_aleatorio, marker="o", label="BST")
ejes[0].plot(tamanos, h_avl_aleatorio, marker="o", label="AVL")
ejes[0].plot(tamanos, referencia, linestyle="--", label="log2(n)")
ejes[0].set_title("Altura con códigos aleatorios")

ejes[1].plot(tamanos, h_bst_ordenado, marker="o", label="BST")
ejes[1].plot(tamanos, h_avl_ordenado, marker="o", label="AVL")
ejes[1].set_title("Altura con códigos ordenados")

for eje in ejes:
    eje.set_xlabel("Cantidad de códigos n")
    eje.set_ylabel("Altura (niveles)")
    eje.grid(True)
    eje.legend()

plt.tight_layout()
plt.show()

for i, n in enumerate(tamanos):
    assert h_bst_ordenado[i] == n
    assert h_avl_ordenado[i] <= 1.45 * math.log2(n + 2)
    assert h_avl_aleatorio[i] <= 1.45 * math.log2(n + 2)

print(f"Con n = {tamanos[-1]}:")
print(f"Altura BST aleatorio: {h_bst_aleatorio[-1]} | Altura AVL aleatorio: {h_avl_aleatorio[-1]}")
print(f"Altura BST ordenado: {h_bst_ordenado[-1]} | Altura AVL ordenado: {h_avl_ordenado[-1]}")
```

```python
# ==========================================================
# PRUEBAS DE RENDIMIENTO: RECORRIDOS
# ==========================================================

tamanos_recorrido = list(range(2000, 20001, 2000))
t_pre, t_in, t_post = [], [], []

for n in tamanos_recorrido:
    datos = random.sample(range(1, n * 10), n)
    arbol = construir(datos, insertar_avl)

    t, lista = medir(preorden, arbol)
    t_pre.append(t)
    assert len(lista) == n

    t, lista = medir(inorden, arbol)
    t_in.append(t)
    assert lista == sorted(datos)

    t, lista = medir(postorden, arbol)
    t_post.append(t)
    assert len(lista) == n

plt.figure(figsize=(9, 5))
plt.plot(tamanos_recorrido, t_pre, marker="o", label="Preorden")
plt.plot(tamanos_recorrido, t_in, marker="o", label="Inorden")
plt.plot(tamanos_recorrido, t_post, marker="o", label="Postorden")
plt.title("Tiempo de los recorridos en el AVL")
plt.xlabel("Cantidad de nodos n")
plt.ylabel("Tiempo (ms)")
plt.grid(True)
plt.legend()
plt.show()

print(f"Con n = {tamanos_recorrido[-1]}:")
print(f"Preorden: {t_pre[-1]:.2f} ms | Inorden: {t_in[-1]:.2f} ms | Postorden: {t_post[-1]:.2f} ms")
print("\nTodas las pruebas de rendimiento fueron aprobadas.")
```

### Resultados de las pruebas de rendimiento

- Con códigos aleatorios ambos árboles crecen de forma parecida a n log n. El BST es más rápido en la práctica porque no calcula alturas ni hace rotaciones, pero su altura quedó casi en el doble de la del AVL (23 contra 12 niveles con n = 1500).
- Con códigos ordenados el BST se degenera en una lista: su altura es exactamente n y el tiempo de construcción crece de forma cuadrática, como predice T(n) = O(n²). En esta ejecución, con n = 1500 tardó alrededor de 20 veces más que el AVL.
- El AVL mantiene una altura cercana a log2(n) sin importar el orden de los datos, lo que confirma que su construcción es O(n log n) en todos los casos.
- Los tres recorridos tardan prácticamente lo mismo y su tiempo crece en línea recta con n, lo que confirma que son O(n).

Los tiempos exactos cambian según el computador, pero la forma de las curvas se mantiene.

## Conclusiones

1. El BST depende del orden en que llegan los datos y puede quedar con una altura grande.
2. El AVL utiliza rotaciones para mantener el factor de balance entre -1 y 1.
3. Ambos árboles contienen los mismos códigos, pero pueden tener estructuras diferentes.
4. El recorrido inorden de un BST entrega los valores ordenados de menor a mayor.
5. Los recorridos preorden, inorden y postorden tienen complejidad O(n), porque visitan cada nodo una vez.
